# How to find midpoints of region

For plotly hover in particular

## Code setup

```python
import os
from dataclasses import dataclass

import numpy as np  # to load grids
import json  # to load dict
import pandas as pd

# Add path to package:
import sys
sys.path.insert(0,'../')
import isopods.grid
import isopods.convert
```

```python
@dataclass(frozen=True)
class Paths:
    '''Singleton object for storing paths to data and database.'''

    dir_data: str = '../data'
    dict_filename: str = 'lsoa2011_devon_1k_grid_dict.json'
    district_grid_data_filename: str = 'district2011_devon_1k_grid_data.csv'
    district_grid_mask_filename: str = 'district2011_devon_1k_grid_mask.csv'

paths = Paths()
```

## Load data

```python
# Grid dictionary
p = os.path.join(paths.dir_data, paths.dict_filename)
with open(p, 'r') as f:
    transform_dict = json.load(f)

# Grid array
p_d = os.path.join(paths.dir_data, paths.district_grid_data_filename)
p_m = os.path.join(paths.dir_data, paths.district_grid_mask_filename)
grid_data = np.loadtxt(p_d, dtype=str, delimiter=',')
grid_mask = np.loadtxt(p_m, dtype=bool, delimiter=',')
arr = np.ma.MaskedArray(
    grid_data, mask=grid_mask, dtype=object)
```

Regions present:

```python
try:
    labels_unique = np.unique(arr.compressed())  # masked array
except AttributeError:
    labels_unique = np.unique(arr)  # normal array
```

```python
labels_unique
```

```python
coords_pix = isopods.convert.pick_midpoints_for_regions(
    arr, labels_unique)
coords = isopods.grid.transform_coords_to_grid(
    coords_pix, **transform_dict)

df_meds = pd.DataFrame(
    coords,
    index=pd.Series(labels_unique, name='region'),
    columns=['x_mid', 'y_mid']
)
```
