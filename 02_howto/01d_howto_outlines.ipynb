# How to make region outlines

Draw outlines around regions in the grid space

```python
# Setting up paths to files:
import os
from dataclasses import dataclass

# Loading data:
import numpy as np  # for loading grids
import json  # for loading grid dictionary

# Add path to package:
import sys
sys.path.insert(0,'../')
import isopods.outline
```

```python
@dataclass(frozen=True)
class Paths:
    '''Singleton object for storing paths to data and database.'''

    dir_data: str = '../data'
    dict_filename: str = 'lsoa2011_devon_1k_grid_dict.json'
    district_grid_data_filename: str = 'district2011_devon_1k_grid_data.csv'
    district_grid_mask_filename: str = 'district2011_devon_1k_grid_mask.csv'

paths = Paths()
```

## Set up grid

```python
p = os.path.join(paths.dir_data, paths.dict_filename)
with open(p, 'r') as f:
    transform_dict = json.load(f)
```

```python
p = os.path.join(paths.dir_data, paths.district_grid_data_filename)
district_grid_data = np.loadtxt(p, dtype=str, delimiter=',')
p = os.path.join(paths.dir_data, paths.district_grid_mask_filename)
district_grid_mask = np.loadtxt(p, dtype=bool, delimiter=',')

arr_districts = np.ma.MaskedArray(
    district_grid_data, mask=district_grid_mask, dtype=object)
```

Make outlines:

```python
dict_polys_pixels = isopods.outline.convert_grid_to_polys(arr_districts)
dict_polys = isopods.outline.transform_poly_dict_coords(
    dict_polys_pixels, transform_dict)
```

Note that the original output is in pixel coordinates. The transform function turns these into coordinates to match the grid space.
